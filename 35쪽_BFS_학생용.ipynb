#35쪽 너비 우선 탐색 [스스로 해결하기] QR코드 페이지

* 아래 예시를 살펴보고, 교과서 34쪽 [그림 I-12]의 트리를 코드로 구현해 보자.

```python
from collections import deque

def BFS(graph, start):
    # 시작 노드를 방문했음을 표시
    visited = {node: False for node in graph}
    queue = deque([start])
    visited[start] = True

    while queue:
        node = queue.popleft()
        print(node, end=' ')

        # 시작 노드와 연결된 모든 미방문 노드를 재귀적으로 탐색
        for neighbor in graph[node]:
            if not visited[neighbor]:
                queue.append(neighbor)
                # 방문 여부를 나타내는 배열
                visited[neighbor] = True
```

# 예시 코드

```python
# 트리 모양 구현
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A'],
    'D': ['B'],
    'E': ['B', 'C']
}
```

```python
# A 노드에서 BFS 시작
BFS(graph, 'A')
```
