#32쪽 깊이 우선 탐색 [스스로 해결하기] QR코드 페이지

* 아래 예시를 살펴보고, 교과서 32쪽 [그림 I-10]의 트리를 코드로 구현해 보자.

tree2.png

```python
# 깊이 우선 탐색 DFS 함수
def DFS(graph, start, visited):
    # 시작 노드를 방문했음을 표시
    visited[start] = True
    print(start, end=' ')

    # 시작 노드와 연결된 모든 미방문 노드를 재귀적으로 탐색
    for neighbor in graph[start]:
        if not visited[neighbor]:
            DFS(graph, neighbor, visited)
```

#  예시 코드

```python
# 트리 모양 구현
tree = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A'],
    'D': ['B'],
    'E': ['B', 'C']
}
```

```python
# 방문 여부를 나타내는 배열
visited = {node: False for node in tree}

# A 노드에서 DFS 시작
DFS(tree, 'A', visited)
```
